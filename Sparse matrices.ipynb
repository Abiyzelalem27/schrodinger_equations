## Sparse matrices

The Hamiltonian matrix $H$ of a physical system is often **sparse**:
most of its matrix elements are zero. In many examples, the number of
nonzero elements grows approximately as $N$, although the full matrix
has $N^2$ positions.

For example, the finite-difference Hamiltonian from Lecture 1 is
tridiagonal. An $N\times N$ tridiagonal matrix has at most $3N-2$
nonzero elements. Storing all its zeros would waste memory.

### Ways to store a sparse matrix

- **Index–value pairs:** Store only each nonzero element and its
  position $(i,j)$, for example $(i,j,A_{ij})$. SciPy calls this
  coordinate format (`coo_matrix`).

- **Compressed sparse row (CSR):** Group the stored nonzero elements
  by row. SciPy provides `csr_matrix` (and the related column format
  `csc_matrix`).

- **Diagonal format:** Store each populated diagonal as a vector.
  For example, store the main diagonal and the two neighboring
  diagonals of a tridiagonal matrix. SciPy provides `dia_matrix`.

**Note:** A sparse matrix should not automatically be converted to a
full matrix before solving a large problem. Specialized methods can
compute a few eigenpairs without finding every eigenpair. One example
is SciPy's `eigsh`.
