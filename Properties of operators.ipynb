### Hermitian matrices

A matrix $A$ is **Hermitian** when it equals its conjugate transpose:

$$
A^\dagger=(A^*)^T=A.
$$

Eigenvalues are real.

If every entry of $A$ is real, then $A^*=A$. In this special case, the Hermitian condition becomes

$$
A^T=A.
$$

A real matrix satisfying this condition is called a **symmetric matrix**.

```python
H = np.array([[1, 1j], [-1j, 2]], dtype=complex)
U = np.array([[0, 1], [-1, 0]], dtype=complex)
A = np.array([[1, 1], [0, 2]], dtype=complex)

for name, matrix in [("H", H), ("U", U), ("A", A)]:
    print(name, "Hermitian:", se.utils.hermitian(matrix), )
```

Examples of Hermitian operators are observables include the Hamiltonian, position, momentum, and spin. The Hamiltonian is itself an observable: it represents energy.

For a two-level atom in the basis $(|g\rangle,|e\rangle)$, an example is

$$
H=\hbar\omega|e\rangle\langle e|
=
\begin{pmatrix}
0 & 0\\
0 & \hbar\omega
\end{pmatrix}.
$$

This matrix is Hermitian because $H^\dagger=H$. Its eigenvalues, $0$ and $\hbar\omega$, are real.

### Unitary matrices

A matrix $A$ is **unitary** if its conjugate transpose is its inverse:

$$
A^\dagger=A^{-1}.
$$

Equivalently,

$$
A^\dagger A=AA^\dagger=I,
$$

Examples of unitary operators are the **time evolution operator**
$e^{-iHt}$ (using $\hbar=1$) and **basis-change operators**.

Unitary operators preserve lengths and angles because they preserve inner products:

$$
\langle A\psi|A\phi\rangle
=
\langle\psi|A^\dagger A|\phi\rangle
=
\langle\psi|\phi\rangle.
$$

A unitary operator has unit eigenvalues $
|\lambda_i|=1.
$.

where $I$ is the identity matrix. For a two-dimensional system,

$$
I=
\begin{pmatrix}
1&0\\
0&1
\end{pmatrix}.
$$
so  $
|\lambda_1|=
$ =  $
|\lambda_2|=1.
$.


For example, $\lambda_i=-1$ or $\lambda_i=i$ is possible because both have magnitude $1$.

```python
for name, matrix in [("H", H), ("U", U), ("A", A)]:
    print(name, "Unitary:", se.utils.unitary(matrix),)
```

### **Normal matrices:** 

A matrix $A$ is normal if it commutes with its Hermitian conjugate:

  $$
  A^\dagger A=AA^\dagger.
  $$

It(normal matrices) is commutes with its adjoint:

$$
[A^\dagger,A]
=A^\dagger A-AA^\dagger
=0.
$$

  Hermitian and unitary matrices are examples of normal matrices.

  Every normal matrix can be diagonalized by a unitary matrix. That is, there exists a unitary matrix $U$ such that

  $$
  U^{-1}AU=D,
  $$

  where $D$ is diagonal. Since $U^{-1}=U^\dagger$, we can also write

  $$
  D=U^\dagger AU.
  $$

```python
for name, matrix in [("H", H), ("U", U), ("A", A)]:
    print(name, "Normal:", se.utils.normal(matrix),)
```
