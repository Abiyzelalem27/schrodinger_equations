# Solving the Infinite Square Well

## Basic Set Up

Suppose an electron (mass $m$) is trapped in an infinite square well; $0 \le x \le a$ such that:
$$
V(x) = 
\begin{cases} 
0 & \text{for } 0 \le x \le a \\
\infty & \text{otherwise.}
\end{cases}
$$

The Time-Independent Schrödinger Equation for this system is:
$$
\hat{H}\psi = E\psi \iff -\frac{\hbar^2}{2m} \frac{d^2\psi}{dx^2} = E\psi
$$

With boundary conditions: $\psi(0) = \psi(a) = 0$ to ensure that the wavefunction is continuous. There is also a normalising constraint:
$$
\int_{-\infty}^{\infty} |\psi_n(x)|^2 dx = 1.
$$

This is because $|\psi|^2$ is the probability of finding the particle between $x$ and $x + dx$ so it must integrate to 1 over all space.

We know that the solutions are just:
$$
\psi_n(x) = \sqrt{\frac{2}{a}} \sin\left(\frac{n\pi x}{a}\right).
$$#

And the energy levels are:
$$
E_n = \frac{\hbar^2 \pi^2 n^2}{2m a^2}.
$$

## Numerical Method

To solve the TISE using Python, we need to turn the Schrödinger differential equation into a *difference equation*.

We discretise space into an $N$ point grid with $x_0 = 0$ and $x_N = a$ and a grid point spacing of $\Delta x$.

The Hamiltonian for the infinite square well is just the kinetic energy: $-\frac{\hbar^2}{2m} \frac{d^2}{dx^2}$.

So we need to express the differential operator $\frac{d^2}{dx^2}$ on our grid!

First, we can approximate the derivative of $f$ at $x_i$ by a *finite difference*:
$$
f'(x_i) \approx \frac{f(x_i + \Delta x) - f(x_i)}{\Delta x} = \frac{f(x_{i+1}) - f(x_i)}{\Delta x}.
$$

It is important to note that this quantity is not defined at the end of the grid since $x_{N+1}$ does not exist. The above expression is called a *forward difference*. We can also have backwards and central differences. 



**Figure credit:** [Finite Difference Method — Wikimedia Commons](https://commons.wikimedia.org/wiki/File:Finite_difference_method.svg)

Using all of this, we can approximate $f''(x_i)$. First, we express $f''(x_i)$ as a forward difference of first derivatives:
$$
f''(x_i) \approx \frac{f'(x_{i+1}) - f'(x_i)}{\Delta x}
$$

Then, each derivative in the expression above can be written as a backward difference, so that:
$$
f''(x_i) \approx \frac{\frac{f(x_{i+1}) - f(x_i)}{\Delta x} - \frac{f(x_i) - f(x_{i-1})}{\Delta x}}{\Delta x} = \frac{f(x_{i+1}) - 2f(x_i) + f(x_{i-1})}{\Delta x^2}
$$

So the differential operator $\frac{d^2}{dx^2}$ is a linear operator (matrix) on our discrete space (check the matrix vector product!). Now we can write the TISE in discrete form:
$$
\hat{H} = -\frac{\hbar^2}{2m} \frac{d^2 \psi}{dx^2} \approx -\frac{\hbar^2}{2m} \frac{1}{(\Delta x)^2}
\begin{pmatrix}
-2 & 1 & 0 & 0 & \dots \\
1 & -2 & 1 & 0 & \dots \\
0 & 1 & -2 & 1 & \dots \\
\vdots & \ddots & \ddots & \ddots & \vdots \\
0 & \dots & \dots & 0 & 1 & -2
\end{pmatrix}
\begin{pmatrix}
\psi(x_1) \\
\psi(x_2) \\
\vdots \\
\vdots \\
\psi(x_{N-1})
\end{pmatrix}
$$

Note we only include steps from $\psi(x_1)$ to $\psi(x_{N-1})$ because of the boundary conditions $\psi(0) = \psi(a) = 0$.

## Integral

In a similar way, we can define an integral operator that integrates a function on our discrete space. This can just be expressed by a *Riemann Sum*:
$$
\int_{x_1}^{x_N} f(x)dx \approx \sum_{i=1}^{N} f(x_i)\Delta x
$$

Now we have all we need to solve the infinite square well problem using Python!

# Steps Solving the Time-Independent Schrödinger Equation (TISE) Numerically

To solve the TISE using Python, we turn the continuous Schrödinger differential equation into a discrete **difference equation**.

## 1. Discretizing the Spatial Grid

Instead of a continuous spatial coordinate $x$, we divide the region $[0, a]$ into an $N+1$ step grid with spacing $\Delta x = \frac{a}{N+1}$:

* **Grid points:** $x_0 = 0, \, x_1 = \Delta x, \, x_2 = 2\Delta x, \, \dots, \, x_{N+1} = a$
* **Boundary conditions:** $\psi(x_0) = \psi(0) = 0$ and $\psi(x_{N+1}) = \psi(a) = 0$
* **Interior points vector:** $\boldsymbol{\psi} = [\psi_1, \psi_2, \dots, \psi_N]^T$

## 3. Matrix Formulation ($\hat{H}\boldsymbol{\psi} = E\boldsymbol{\psi}$)

Applying this equation across all interior points ($i = 1, 2, \dots, N$) yields a tridiagonal matrix eigenvalue problem:

### How to Solve:
* **Analytical / Manual:** Solve the determinant equation $\det(H - EI) = 0$ for small $N$.
* **Python Numerical:** Construct matrix $H$ using `numpy` and solve for eigenvalues $E$ and eigenvectors $\boldsymbol{\psi}$ using `numpy.linalg.eigh(H)`.

```python
# import libirary 
import numpy as np
import matplotlib.pyplot as plt
import se # local package
```

```python
# create the grid 
a = 2
N = 600
x = np.linspace(0, a, N + 1)
dx = x[1]-x[0]

# constant AMU
hbar = 1
m = 1
```

```python
# create the differential operators 
M = np.diag(-2*np.ones(N-1)) + np.diag(np.ones(N-2), 1) + np.diag(np.ones(N-2), -1)
print(M)
print("")
print(M.shape)
```

```python
# Hamiltonian
H = -hbar**2/(2 * m) * 1/(dx)**2 * M

# solve the Equations
E, psi = np.linalg.eigh(H)
```

```python
psi = psi.T
```

```python
norm = se.utils.integral(psi**2, dx, axis=0)
psi = psi/np.sqrt(norm)
```

```python
plt.figure(figsize=(10, 8))

for n in range(0, 6):
    plt.plot(x[1:-1], psi[n]**2, label='psi^2_{}'.format(n))
    plt.legend()
```

## Explaining the Probability Density Graphs $|\psi(x)|^2$

### Key Takeaways

* **First 6 States:** The plot displays the first 6 quantum energy states ($n = 1$ to $n = 6$), corresponding to `psi^2_0` through `psi^2_5` in 0-indexed Python notation.
* **Number of Peaks = State Number ($n$):** Each probability density $|\psi_n(x)|^2$ contains exactly $n$ distinct peaks:
  * $n = 1$ (`psi^2_0`): **1 central peak** (Ground State)
  * $n = 2$ (`psi^2_1`): **2 peaks**
  * $n = 3$ (`psi^2_2`): **3 peaks**
  * $n = 4$ (`psi^2_3`): **4 peaks**
  * $n = 5$ (`psi^2_4`): **5 peaks**
  * $n = 6$ (`psi^2_5`): **6 peaks**

---

### Physical Interpretation

1. **Nodes (Zero Probability Points):** The points where $|\psi(x)|^2 = 0$ between peaks inside the well are called **nodes**. A wavefunction of state $n$ has exactly $n - 1$ internal nodes where the particle has zero probability of being found.
2. **Boundary Conditions:** All curves drop to $0$ at $x = 0$ and $x = a$ because the walls of the infinite square well are infinitely hard barriers, forcing $\psi(0) = \psi(a) = 0$.
3. **Probability Normalization:** The total area under each $|\psi_n(x)|^2$ curve equals $1$ ($\int_0^a |\psi_n(x)|^2 dx = 1$), representing a $100\%$ probability of finding the particle inside the well.
